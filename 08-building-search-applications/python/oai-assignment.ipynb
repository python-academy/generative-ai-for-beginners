```python
# python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.core import Settings

Settings.embed_model = HuggingFaceEmbedding(
    model_name="BAAI/bge-small-en-v1.5"
)
```

```python
text = 'the quick brown fox jumped over the lazy dog'

embedding = HuggingFaceEmbedding().get_text_embedding(text)

print(type(embedding))
print(embedding)  # print first 5 dimensions
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
# compare several words
automobile_embedding    = HuggingFaceEmbedding().get_text_embedding('automobile')
vehicle_embedding       = HuggingFaceEmbedding().get_text_embedding('vehicle')
dinosaur_embedding      = HuggingFaceEmbedding().get_text_embedding('dinosaur')
stick_embedding         = HuggingFaceEmbedding().get_text_embedding('stick')

# comparing cosine similarity, automobiles vs automobiles should be 1.0, i.e exactly the same, while automobiles vs dinosaurs should be between 0 and 1, i.e. not the same
print(cosine_similarity(automobile_embedding, automobile_embedding))
print(cosine_similarity(automobile_embedding, vehicle_embedding))
print(cosine_similarity(automobile_embedding, dinosaur_embedding))
print(cosine_similarity(automobile_embedding, stick_embedding))
```
